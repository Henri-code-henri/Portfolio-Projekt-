```python
import pandas as pd
```

```python
stations = pd.read_csv("data/raw/stations_metadaten.csv")
stations.head()
```

```python
stations.info()
```

```python
tirol_stations = stations[stations["Bundesland"] == "Tirol"]

tirol_stations[[
    "id",
    "Stationsname",
    "Höhe [m]",
    "Startdatum",
    "Enddatum"
]].head(10)
```

```python
klima = pd.read_csv("data/raw/Tirol_Stationsdaten-v2 (1 m)_ Qualitätsgeprüfte Stationsdaten.csv")

klima.head()
```

```python
klima.info()
```

```python
klima["station"].nunique()
```

```python
klima = klima.merge(
    tirol_stations[["id", "Stationsname", "Höhe [m]", "Startdatum", "Enddatum"]],
    left_on="station",
    right_on="id",
    how="left"
)

klima.head()
```

```python
klima["Stationsname"].isna().sum()
```

```python
temperatur_check = klima.groupby(
    ["station", "Stationsname", "Höhe [m]"]
)["tl_mittel"].count().sort_values(ascending=False)

temperatur_check.head(20)
```

```python
parameter = [
    "tl_mittel",
    "tlmax_mittel",
    "tlmin_mittel",
    "tage_frost",
    "shneu_manu",
    "tage_schdecke",
    "sh_manu_max",
    "rr",
    "tage_festrrp"
]

daten_check = klima.groupby(
    ["station", "Stationsname", "Höhe [m]"]
)[parameter].count()

daten_check.head(20)
```

```python
temp_zeitraum = (
    klima.dropna(subset=["tl_mittel"])
    .groupby(["station", "Stationsname", "Höhe [m]"])["time"]
    .agg(["min", "max", "count"])
    .sort_values("count", ascending=False)
)

temp_zeitraum.head(20)
```

```python
schnee_zeitraum = (
    klima.dropna(subset=["tage_schdecke"])
    .groupby(["station", "Stationsname", "Höhe [m]"])["time"]
    .agg(["min", "max", "count"])
    .sort_values("count", ascending=False)
)

schnee_zeitraum.head(20)
```

```python
klima["time"] = pd.to_datetime(klima["time"], utc=True)

analyse = klima[
    (klima["time"] >= "1970-01-01") &
    (klima["time"] <= "2025-12-31")
].copy()

analyse.shape
```

```python
vollstaendigkeit = analyse.groupby(
    ["station", "Stationsname", "Höhe [m]"]
)[parameter].count()

vollstaendigkeit.head(20)
```

```python
vollstaendigkeit_prozent = (vollstaendigkeit / 672 * 100).round(1)

vollstaendigkeit_prozent.head(20)
```

```python
geeignete_stationen = vollstaendigkeit_prozent[
    (vollstaendigkeit_prozent["tl_mittel"] >= 90) &
    (vollstaendigkeit_prozent["tage_schdecke"] >= 90)
]

geeignete_stationen
```

```python
def hoehenklasse(hoehe):
    if hoehe < 800:
        return "Tal (<800 m)"
    elif hoehe <= 1300:
        return "Mittellage (800-1300 m)"
    else:
        return "Hochlage (>1300 m)"

analyse["Höhenklasse"] = analyse["Höhe [m]"].apply(hoehenklasse)
```

```python
station_ids = geeignete_stationen.index.get_level_values("station")

tirol_final = analyse[
    analyse["station"].isin(station_ids)
].copy()

tirol_final.shape
```

```python
tirol_final[parameter].isna().sum()
```

```python
tirol_final["Jahr"] = tirol_final["time"].dt.year
tirol_final["Monat"] = tirol_final["time"].dt.month

tirol_final[["time", "Jahr", "Monat"]].head()
```

```python
jahresdaten = tirol_final.groupby(
    ["station", "Stationsname", "Höhe [m]", "Höhenklasse", "Jahr"]
).agg({
    "tl_mittel": "mean",
    "tlmax_mittel": "mean",
    "tlmin_mittel": "mean",
    "tage_frost": "sum",
    "shneu_manu": "sum",
    "tage_schdecke": "sum",
    "sh_manu_max": "max",
    "rr": "sum"
}).reset_index()

jahresdaten.head()
```

```python
monate_pro_jahr = tirol_final.groupby(
    ["station", "Stationsname", "Jahr"]
)[parameter].count()

monate_pro_jahr.head(20)
```

```python
jahresdaten = tirol_final.groupby(
    ["station", "Stationsname", "Höhe [m]", "Höhenklasse", "Jahr"]
).agg(
    tl_mittel=("tl_mittel", "mean"),
    tlmax_mittel=("tlmax_mittel", "mean"),
    tlmin_mittel=("tlmin_mittel", "mean"),
    tage_frost=("tage_frost", lambda x: x.sum(min_count=10)),
    shneu_manu=("shneu_manu", lambda x: x.sum(min_count=10)),
    tage_schdecke=("tage_schdecke", lambda x: x.sum(min_count=10)),
    sh_manu_max=("sh_manu_max", "max"),
    rr=("rr", lambda x: x.sum(min_count=10))
).reset_index()

jahresdaten.head(10)
```

```python
analyse_1940 = klima[
    (klima["time"] >= "1940-01-01") &
    (klima["time"] <= "2025-12-31")
].copy()

monate_1940 = 86 * 12

check_1940 = analyse_1940.groupby(
    ["station", "Stationsname", "Höhe [m]"]
)[["tl_mittel", "tage_schdecke"]].count()

check_1940_prozent = (check_1940 / monate_1940 * 100).round(1)

geeignet_1940 = check_1940_prozent[
    (check_1940_prozent["tl_mittel"] >= 90) &
    (check_1940_prozent["tage_schdecke"] >= 90)
]

geeignet_1940
```
